### Extracting all the movies no tv shows

```python
movies = my_df[my_df["type"] == "Movie"]
```

### Cleaning the netflix dataset

```python
movies_copy = movies.copy()
```

### Checking missing values

```python
movies_copy.isnull().sum()
```

### Fixing the missing values

```python
text_columns = [
    "director",
    "cast",
    "country",
    "listed_in",
    "description",
]

for column in text_columns:
    movies_copy[column] = movies_copy[column].fillna("")
movies_copy["rating"] = movies_copy["rating"].fillna("Unknown")
movies_copy["duration"] = movies_copy["duration"].fillna("Unknown")
print(movies_copy[["rating", "duration"]].isnull().sum())
```

### Removing unwanted columns

```python
movies_copy = movies_copy[
    [
        "show_id",
        "title",
        "director",
        "cast",
        "country",
        "release_year",
        "rating",
        "duration",
        "listed_in",
        "description"
    ]
].copy()
```

```python
movies_copy.head()
```

### Combined features 
Director+cast+country+genres+Description

```python
movies_copy["combined_features"] = (
    movies_copy["director"] + " " +
    movies_copy["cast"] + " " +
    movies_copy["country"] + " " +
    movies_copy["listed_in"] + " " +
    movies_copy["description"]
)
```

```python
movies_copy[["title", "combined_features"]].head()
```

```python
movies_copy["combined_features"] = movies_copy["combined_features"].str.lower()
```

```python
movies_copy["combined_features"] = (
    movies_copy["combined_features"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

### saving out cleared dataset

```python
movies_copy.to_csv(
    "../data/netflix_movies_cleaned.csv",
    index=False
)
```
